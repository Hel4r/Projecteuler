Problem 33

The fraction $49/98$ is a curious fraction, as an inexperienced mathematician in attempting to simplify it may incorrectly believe that $49/98 = 4/8$, which is correct, is obtained by cancelling the $9$s.
We shall consider fractions like, $30/50 = 3/5$, to be trivial examples.
There are exactly four non-trivial examples of this type of fraction, less than one in value, and containing two digits in the numerator and denominator.
If the product of these four fractions is given in its lowest common terms, find the value of the denominator.


Więc tak: sprawdzam liczby od 10 do 99, jeżeli ich iloraz jest taki jak iloraz ze skreślonymi powtarzającymi się cyframi, to dodaję do list licznik i mianownik. 
Najpierw biorę na tape sytuację, gdzie cyfra jedności licznika skraca cyfrę dziesiątek mianownika. 

```python
nums = []
dens = []
for n in range(10, 99):
    for d in range(n+1, 100):
        n1 = str(n)[0]
        n2 = str(n)[1]
        d1 = str(d)[0]
        d2 = str(d)[1]
        if n2 == d1 and d2 != '0' and n/d == int(n1)/int(d2):
            print(f'{n}/{d}')
            nums.append(n1)
            dens.append(d2)
print([int(n) for n in nums])
print([int(d) for d in dens])
```

Problem 37

The number $3797$ has an interesting property. Being prime itself, it is possible to continuously remove digits from left to right, and remain prime at each stage: $3797$, $797$, $97$, and $7$. Similarly we can work from right to left: $3797$, $379$, $37$, and $3$.
Find the sum of the only eleven primes that are both truncatable from left to right and right to left.

NOTE: $2$, $3$, $5$, and $7$ are not considered to be truncatable primes.


```python
def prime(n):
    if n < 2: return False
    for k in range(2, int(n**(0.5))+1):
        if n % k == 0:
            return False
    return True

def problem_37():
    liczby = []
    k = 10
    while(len(liczby) < 11):
        bool = True
        if prime(k) == False:
            bool = False
        else:
            for i in range(1, len(str(k))):
                if prime(int(str(k)[i:])) == False or prime(int(str(k)[:-i])) == False:
                    bool = False
                    break

        if bool == True:
            liczby.append(k)
        k+=1

    return(sum(liczby))
```

Problem 39

If $p$ is the perimeter of a right angle triangle with integral length sides, $\{a, b, c\}$, there are exactly three solutions for $p = 120$.
$\{20,48,52\}$, $\{24,45,51\}$, $\{30,40,50\}$
For which value of $p \le 1000$, is the number of solutions maximised?


Muszę określić granice długości dla przeciwprostokątnej: nie może być dłuższa niż podłoga połowy obwodu. Nie może być krótsza niż (podłoga ?) z obwodu razy ((2**0.5)/(2 + 2**0.5) = 0.4142135623730951). Przechowuję liczbę możliwych trójkątów i obwód, zmieniam kiedy trzeba. 

prostokątny równoramienny o długościach 4, 4, 5.656854249492381

```python
const = 0.4142135623730951
p_max = 0
triangles_max = 0
for p in range(10, 1001):
    curr_triangles = 0
    for c in range(int(p*const), int(p/2)+1):
        for b in range(math.ceil((p-c)/2), p - c):
            if (p-c-b)**2 + b**2 == c**2:
                curr_triangles += 1
                print(f'obwód: {p}, boki: {[c, b, p-c-b]}')
    if curr_triangles > triangles_max:
        p_max = p
        triangles_max = curr_triangles
print([p_max, triangles_max])
```
