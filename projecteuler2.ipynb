Problem 33

The fraction $49/98$ is a curious fraction, as an inexperienced mathematician in attempting to simplify it may incorrectly believe that $49/98 = 4/8$, which is correct, is obtained by cancelling the $9$s.
We shall consider fractions like, $30/50 = 3/5$, to be trivial examples.
There are exactly four non-trivial examples of this type of fraction, less than one in value, and containing two digits in the numerator and denominator.
If the product of these four fractions is given in its lowest common terms, find the value of the denominator.


Więc tak: sprawdzam liczby od 10 do 99, jeżeli ich iloraz jest taki jak iloraz ze skreślonymi powtarzającymi się cyframi, to dodaję do list licznik i mianownik. 
Najpierw biorę na tape sytuację, gdzie cyfra jedności licznika skraca cyfrę dziesiątek mianownika. 

```python
nums = []
dens = []
for n in range(10, 99):
    for d in range(n+1, 100):
        n1 = str(n)[0]
        n2 = str(n)[1]
        d1 = str(d)[0]
        d2 = str(d)[1]
        if n2 == d1 and d2 != '0' and n/d == int(n1)/int(d2):
            print(f'{n}/{d}')
            nums.append(n1)
            dens.append(d2)
print([int(n) for n in nums])
print([int(d) for d in dens])
```

Problem 37

The number $3797$ has an interesting property. Being prime itself, it is possible to continuously remove digits from left to right, and remain prime at each stage: $3797$, $797$, $97$, and $7$. Similarly we can work from right to left: $3797$, $379$, $37$, and $3$.
Find the sum of the only eleven primes that are both truncatable from left to right and right to left.

NOTE: $2$, $3$, $5$, and $7$ are not considered to be truncatable primes.


```python
def prime(n):
    if n < 2: return False
    for k in range(2, int(n**(0.5))+1):
        if n % k == 0:
            return False
    return True

def problem_37():
    liczby = []
    k = 10
    while(len(liczby) < 11):
        bool = True
        if prime(k) == False:
            bool = False
        else:
            for i in range(1, len(str(k))):
                if prime(int(str(k)[i:])) == False or prime(int(str(k)[:-i])) == False:
                    bool = False
                    break

        if bool == True:
            liczby.append(k)
        k+=1

    return(sum(liczby))
```

Problem 39

If $p$ is the perimeter of a right angle triangle with integral length sides, $\{a, b, c\}$, there are exactly three solutions for $p = 120$.
$\{20,48,52\}$, $\{24,45,51\}$, $\{30,40,50\}$
For which value of $p \le 1000$, is the number of solutions maximised?


Muszę określić granice długości dla przeciwprostokątnej: nie może być dłuższa niż podłoga połowy obwodu. Nie może być krótsza niż (podłoga ?) z obwodu razy ((2**0.5)/(2 + 2**0.5) = 0.4142135623730951). Przechowuję liczbę możliwych trójkątów i obwód, zmieniam kiedy trzeba. 

prostokątny równoramienny o długościach 4, 4, 5.656854249492381

```python
const = 0.4142135623730951
p_max = 0
triangles_max = 0
for p in range(10, 1001):
    curr_triangles = 0
    for c in range(int(p*const), int(p/2)+1):
        for b in range(math.ceil((p-c)/2), p - c):
            if (p-c-b)**2 + b**2 == c**2:
                curr_triangles += 1
                print(f'obwód: {p}, boki: {[c, b, p-c-b]}')
    if curr_triangles > triangles_max:
        p_max = p
        triangles_max = curr_triangles
print([p_max, triangles_max])
```

Problem 27 ZROBIONY

Euler discovered the remarkable quadratic formula:

$$n^2 + n + 41$$

It turns out that the formula will produce **40 primes** for consecutive integer values $0 \le n \le 39$.

However, when $n = 40$:

$$40^2 + 40 + 41 = 40(40 + 1) + 41$$

which is divisible by $41$.

And when $n = 41$:

$$41^2 + 41 + 41$$

is clearly divisible by $41$.

---

The incredible formula:

$$n^2 - 79n + 1601$$

produces **80 primes** for consecutive values $0 \le n \le 79$.

The product of the coefficients $-79$ and $1601$ is:

$$-126479$$

---

Considering quadratics of the form:

> $$n^2 + an + b$$
> where $|a| < 1000$ and $|b| \le 1000$
>
> where $|n|$ is the absolute value of $n$
> e.g. $|11| = 11$ and $|-4| = 4$

---

**Task:**
Find the product of the coefficients $a$ and $b$ for the quadratic expression that produces the maximum number of primes for consecutive values of $n$, starting with $n = 0$.


Wcześniejsza funkcja prime() szybko rozpoznaje liczby ujemne, więc dobrze. 
robię pętlę w pętli (4 mln wykonań). Potrzebuję 3 zmiennych tymczasowych (na aktualne współczynniki i długość ciągu liczb pierwszych) i 3 zmiennych stałych (na najlepsze współczynniki i długość ciągu liczb pierwszych). W każdym wykonaniu pętli wrzucam kolejne liczby naturalne aż któraś wartość nie będzie pierwsza. Jeśli aktualna długość ciągu jest większa od najdłuższego ciągu, to zmieniam, na końcu programu zwracam iloczyn a*b. 

```python
best_a = -999
best_b = -999
longest_seq = 0

for a in range(-999, 1000):
    for b in range(-999, 1000):
        seq = 0 
        v = lambda n: int(n**2 + a*n + b)
        n = 0
        while True:
            if prime(v(n)) == False:
                break
            else:
                seq += 1 # jeśli wartość dla tego n jest pierwsza to zwiększ długość aktualnego ciągu
                n += 1
        if seq > longest_seq:
            best_a = a
            best_b = b
            longest_seq = seq
print([best_a, best_b, longest_seq])
print(f'Odpowiedź: {best_a*best_b}')
```

mogę użyć metody itertools.permutations()

```python
def parzyste_cyfry(n): # zakładam że przyjmuje int
    n = str(n)
    for i in n: # tutaj i to str!
        if int(i) % 2 == 0 or i == '5':
            return True
    return False
print([parzyste_cyfry(k) for k in [2, 3, 11, 27, 747]])
```

```python
pierwsze_nieparzyste_cyfry = [2, 3, 5] + [p for p in range(6, 10**6) if prime(p) and parzyste_cyfry(p) == False]
print(pierwsze_nieparzyste_cyfry)
```

Warto zauważyć, że tutaj tablica 'pierwsze_nieparzyste_cyfry' zawiera liczby pierwsze od 2 do 10**6-1 nie licząc tych, które mają cyfry parzyste lub 5 (2 i 5 należy do tej tablicy). 
Czy mogę "zmergować" tupletkę zawierającą cyfry jako znaki?

```python
licznik = 0
def tupl_to_int(tupletka):
    string = ''
    for i in tupletka:
        string = string + i
    return(int(string))

for p in pierwsze_nieparzyste_cyfry:
    bool = True
    for perm in itertools.permutations(str(p)): # perm to tupletka
        if prime(tupl_to_int(perm)) == False:
            bool = False
            break
    if bool == True:
        licznik += 1
        print(p)
print(f'Odpowiedź: {licznik}')
```
