```python
def max_factor(n):
    f = []
    while n>1:
        for i in range(2,int(np.ceil(np.sqrt(n)))):
            if n % i == 0:
                f.append(i)
                n/=i
            if i == int(np.ceil(np.sqrt(n))):
                f.append(n)
                n = 1
                break
    return max(f)
print(max_factor(600851475143))
```

A Pythagorean triplet is a set of three natural numbers, $a \lt b \lt c,$ for which,
$$a^2 + b^2 = c^2.$$
For example, $3^2 + 4^2 = 9 + 16 = 25 = 5^2.$
There exists exactly one Pythagorean triplet for which $a + b + c = 1000.$ Find the product $abc.$


```python
def is_pythagorean(nums):
    # if(len[nums]) != 3:
    #     print("Nie są trzema liczbami")
    #     return False
    nums = sorted(nums)
    if np.power(nums[0],2) + np.power(nums[1],2) == np.power(nums[2],2):
        return True
    else: 
        return False
    
def needed_numbers():
    for a in range(1,1001):
        for b in range(1,a+1):
            for c in range(1, b+1):
                if a + b + c == 1000 and is_pythagorean([a,b,c]):
                    return [a,b,c]
print(np.product(needed_numbers()))
```

The following iterative sequence is defined for the set of positive integers:
<ul style="list-style-type:none;">
<li>$n \to n/2$ ($n$ is even)</li>
<li>$n \to 3n + 1$ ($n$ is odd)</li></ul>
Using the rule above and starting with $13$, we generate the following sequence:
$$13 \to 40 \to 20 \to 10 \to 5 \to 16 \to 8 \to 4 \to 2 \to 1.$$
It can be seen that this sequence (starting at $13$ and finishing at $1$) contains $10$ terms. Although it has not been proved yet (Collatz Problem), it is thought that all starting numbers finish at $1$.
Which starting number, under one million, produces the longest chain?
 class="note"><b>NOTE:</b> Once the chain starts the terms are allowed to go above one million.


```python
def collatz(n): # działa
    nums = [n]
    length = 1
    while n > 1:
        if n % 2 == 0:
            n/=2
            length += 1
            nums.append(int(n))
        else:
            n = 3 * n + 1
            length += 1
            nums.append(int(n))
    return length 

max_n = 1
max_length = 1
for i in range(10**6):
    if collatz(i) > max_length:
        max_n = i
        max_length = collatz(i)
print(max_n)
```

$$\begin{align}

8208 = 8^4 + 2^4 + 0^4 + 8^4\\
9474 = 9^4 + 4^4 + 7^4 + 4^4
\end{align}$$
As $1 = 1^4$ is not a sum it is not included.

```python
def is_power(n, k): # działa
    n_str = str(n)
    numbers = np.array([int(i) for i in n_str])
    if n == sum(numbers ** k):
        return True
    else:
        return False
    
k = symbols('k')
f = 10**k - 59049*k
print(nsolve(f, 10)) # szukam górnego ograniczenia liczby cyfr, dam 6 cyfr

lista = []
for i in range(2, 999999):
    if is_power(i, 5):
        lista.append(i)
print(sum(np.array(lista)))
# print(lista)
```

Evaluate the sum of all the amicable numbers under $10000$.

Problem 23

A perfect number is a number for which the sum of its proper divisors is exactly equal to the number. For example, the sum of the proper divisors of $28$ would be $1 + 2 + 4 + 7 + 14 = 28$, which means that $28$ is a perfect number.
A number $n$ is called deficient if the sum of its proper divisors is less than $n$ and it is called abundant if this sum exceeds $n$.

As $12$ is the smallest abundant number, $1 + 2 + 3 + 4 + 6 = 16$, the smallest number that can be written as the sum of two abundant numbers is $24$. By mathematical analysis, it can be shown that all integers greater than $28123$ can be written as the sum of two abundant numbers. However, this upper limit cannot be reduced any further by analysis even though it is known that the greatest number that cannot be expressed as the sum of two abundant numbers is less than this limit.
Find the sum of all the positive integers which cannot be written as the sum of two abundant numbers.


```python
# Program powinien działać, ale złożoność O(n^3) skutkuje bardzo wolnym działaniem, trzeba innego podejścia algorytmicznego

def proper_divisors(n):
    divisors = []
    for i in range(1, math.ceil(n/2) + 1):
        if n % i == 0:
            divisors.append(i)
    return divisors

def is_abundant(n):
    if sum(proper_divisors(n)) > n:
        return True
    else:
        return False
    
def is_deficient(n):
    if sum(proper_divisors(n)) < n:
        return True
    else:
        return False
    
def expressable_as_sum_of_abundant(n): # działa
    bool = False
    for i in range(1, math.floor(n / 2) + 1):
        if is_abundant(i) == True and is_abundant(n - i) == True:
            bool = True
            break
    return bool

suma = 0
for i in range(24, 28124):
    if expressable_as_sum_of_abundant(i) == False:
        suma += i
print(suma)
```

If they are then placed in numerical order, with any repeats removed, we get the following sequence of $15$ distinct terms:
$$4, 8, 9, 16, 25, 27, 32, 64, 81, 125, 243, 256, 625, 1024, 3125.$$
How many distinct terms are in the sequence generated by $a^b$ for $2 \le a \le 100$ and $2 \le b \le 100$?

```python
# done

def distinct_terms():
    lista = []
    for a in range(2, 101):
        for b in range(2, 101):
            lista.append(a**b)
    return len(set(lista))

print(distinct_terms())
```

Problem 36

The decimal number, $585 = 1001001001_2$ (binary), is palindromic in both bases.
Find the sum of all numbers, less than one million, which are palindromic in base $10$ and base $2$.
(Please note that the palindromic number, in either base, may not include leading zeros.)


```python
# done

def binary(n):
    cyfry = []
    while n>0:
        cyfry.append(n % 2)
        n = int(n/2)
    return cyfry == cyfry[::-1]

palindromic = lambda a: str(a) == str(a)[::-1]

suma = 0
for i in range(10**6):
    if binary(i) == True and palindromic(i) == True:
        suma += i
print(suma)
print(binary(1))
```

Problem 34

$145$ is a curious number, as $1! + 4! + 5! = 1 + 24 + 120 = 145$.
Find the sum of all numbers which are equal to the sum of the factorial of their digits.
Note: As $1! = 1$ and $2! = 2$ are not sums they are not included.

```python
# done

def is_curious(n):
    k = str(n)
    silnie = [math.factorial(int(liczba)) for liczba in k]
    return n == sum(silnie)

def problem34():
    suma = 0
    for i in range(3, 10**7):
        if is_curious(i):
            suma += i
    return suma

print(problem34())
```

Problem 52

It can be seen that the number, $125874$, and its double, $251748$, contain exactly the same digits, but in a different order.
Find the smallest positive integer, $x$, such that $2x$, $3x$, $4x$, $5x$, and $6x$, contain the same digits.


```python
# done

def permuted(n):
    raz = sorted(str(n))
    dwa = sorted(str(2*n))
    trzy = sorted(str(3*n))
    cztery = sorted(str(4*n))
    piec = sorted(str(5*n))
    szesc = sorted(str(6*n))
    return raz == dwa == trzy == cztery == piec == szesc

def problem52():
    for i in range(1, 10**6):
        if permuted(i):
            return i
            break
print(problem52())
```
